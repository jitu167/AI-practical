```python

import gradio as gr
from datetime import datetime
```

```python
# Chatbot response function
def chatbot_response(message, history):

    user_message = message.lower().strip()

    if "hello" in user_message or "hi" in user_message:
        return "Hello! Welcome to my Gradio chatbot. How can I help you today?"

    elif "your name" in user_message:
        return "I am Gradio AI Assistant."

    elif "time" in user_message:
        return f"The current time is {datetime.now().strftime('%H:%M:%S')}."

    elif "data structure" in user_message:
        return (
            "A data structure is a specialized format for organizing, storing, and managing data in a computer so that it can be accessed and modified efficiently. Choosing the right data structure improves program speed and reduces memoryusage "
        )

    elif "greedy" in user_message:
        return (
            "A greedy algorithm is an approach to solving optimization problems by making the best possible choice at each step "
            "(the locally optimal choice) in hopes of finding the best overall solution (the global optimum)."
            "It commits to each decision immediately and never goes back to reconsider past choices."
        )

    elif "thank" in user_message:
        return "You're welcome! Feel free to ask me anything else."

    else:
        return (
            f"You asked: '{message}'\n\n"
            "That's an interesting question! I am a demo conversational "
            "assistant. "
        )

```

```python
# Create Gradio chatbot interface
demo = gr.ChatInterface(
    fn=chatbot_response,

    title="Gradio Chatbot",

    description=(
        "Welcome to the gradio Chatbot! "
    ),

    examples=[
        "Hello! How are you?",
        "What is Data Structure?",
        "What is Greedy Algorithm?"
    ],
)


# Launch the application
if __name__ == "__main__":
    demo.launch(
        server_name="127.0.0.1",
        server_port=7860,
        share=False
    )
```
