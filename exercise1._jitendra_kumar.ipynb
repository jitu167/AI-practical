```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
df = pd.DataFrame(
    np.random.uniform(10, 100, size=(50, 4)),
    columns=['f1', 'f2', 'f3', 'f4']
).round(2)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['f2'], df['f3'], color='#2b5c8f', alpha=0.8, edgecolors='white', s=65)
plt.title('Relationship Between Column f2 and Column f3', fontsize=13, fontweight='bold')
plt.xlabel('Column f2')
plt.ylabel('Column f3')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1).round(2)
df.to_excel('random_data_analysis.xlsx', index=False)
```
